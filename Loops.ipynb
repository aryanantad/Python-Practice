```python
#WAP to find sum of first n numbers(using while)
n = 7
sum= 0
i = 1
while i <= n:
    sum +=i
    i += 1

print("Total sum :",sum)
```

```python
#WAP to find factorial of first n numbers (use for)
n = 5
fact = 1
for i in range(1, n+1):
    fact *=i
print("factorial value=",fact)
```
