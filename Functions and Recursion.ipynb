```python
#WAF to find factorial of n

def cal_fact(n):
    fact =1
    for i in range(1,n+1):
        fact *= i
    print(fact)
cal_fact(5)
```

```python
#recursion
def show(n):
    if(n==0):
        return
    print(n)
    show(n-1)
show(5)
```

```python
def fact(n):  
    if(n==0 or n==1):
        return 1
    else: 
        return n * fact(n-1)
print (fact(5))
```

```python
# Recursion function to calculate sum of first n numbers
def calc_sum(n):
    if n == 0:
        return 0
    return calc_sum(n - 1) + n

sum = calc_sum(10)
print(sum)
```

```python
#write a recurssive func to print all elements in a list .use list and index as parameters
def print_list(list,idx=0):
    if(idx == len(list)):
        return
    print(list[idx])
    print_list(list,idx+1)
fruits = ["mango","apple","banana"]
print_list(fruits)
```
