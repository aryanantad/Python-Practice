```python
#dict methods
student = {
    "name": "Aryan",
    " score": {
        "chem": 98,
        "math": 86
    }
}
print(student.keys())
print(list(student.values()))
print(student.items())
print(student.get("name"))
print(student.update({"city" : "PUNE"}))
```
