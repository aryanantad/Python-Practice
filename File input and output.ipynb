```python
#create a file practice.txt using pyhton add the data
with open("practice.txt","w") as f:
    f.write("hi everyone \n we are learning file I/O\n")
    f.write("using java \nI like programming")
    
```

```python
#WAF that replaces all occurances of java with python in same above file
# Step 1: Read the file
with open("practice.txt", "r") as f:
    data = f.read()

# Step 2: Replace text
new_data = data.replace("java", "python")
print(new_data)

# Step 3: Write the updated content back
with open("practice.txt", "w") as f:
    f.write(new_data)
```

```python
#From a file containing numbers seperated by comma ,print count of even nos
f = open("practice.txt", "r")
count = 0
with open("practice.txt","r"):
    data = f.read()
    
    nums = data.split(",")
    for val in nums:
        if(int(val) % 2== 0):
            count +=1
print(count)
```
